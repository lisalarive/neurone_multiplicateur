# Petit réseau de neurones – Prédiction d'une fonction mathématique

Projet réalisé en 2023, lorsque j'étais élève au lycée Simone Veil à Noisiel, dans le cadre d'un partenariat avec l'École des Ponts et Chaussées.

Avec l'aide d'un étudiant de l'École des Ponts et Chaussées, j'ai découvert le fonctionnement d'un petit réseau de neurones à travers un exemple simple en Python.

L'objectif est d'entraîner un neurone à reproduire la fonction :

**f(x, y) = 2x + 3y**

À partir de plusieurs données d'entraînement, les poids du neurone sont progressivement ajustés en fonction de l'erreur entre le résultat attendu et le résultat obtenu.

Ce projet constitue une première approche concrète de l'apprentissage automatique et m'a permis de mieux comprendre, par la pratique, le principe d'apprentissage d'un neurone artificiel.

```python
Donnees_Entrainement = [[2,3,13], [3,5,21], [7,8,38], [3,6,24]] # a = 2, b = 3, x, y donne le résultat 2x+3y
# on lui donne des données d'entrainement

class Neurone():
    poids1 = 0
    poids2 = 0

def modif_poids(entree1, entree2, sortie, neurone):
    """ La fonction sert à """
    output_neuron = entree1*neurone.poids1 + entree2*neurone.poids2
    erreur = sortie - output_neuron
    neurone.poids1 += erreur*entree1*0.01
    neurone.poids2 += erreur*entree2*0.01

def test(Donnees_Entrainement, neurone):
    for i in range(len(Donnees_Entrainement)):
        modif_poids(Donnees_Entrainement[i][0], Donnees_Entrainement[i][1], Donnees_Entrainement[i][2], neurone)

def Entrainement_Neurone(data, nb_entrainement):
    neurone = Neurone
    for i in range(nb_entrainement):
        test(data, neurone)
    return neurone

neurone_entraine = Entrainement_Neurone(Donnees_Entrainement, 250000)
print(neurone_entraine.poids1*3 + neurone_entraine.poids2*7)
```
